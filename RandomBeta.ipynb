# Beta distribution

- models probabilities of probabilities

- input space is not infinite as compared to other continuous distributions, it's bounded to [0, 1]

$$P(x; \alpha, \beta) =  \frac{x^{\alpha - 1} (1-x)^{\beta - 1}}{B(\alpha, \beta)}$$

***

### Example

Given a biased coin with unknown probability $p$ of heads we observe the following data: 7 heads, 3 tails. Classical probability approach for finding $p$ would be $p= \frac{7}{10}=0.7$. Since we cannot be certain for this to be the true probability we instead would like to build a distribution where we treat $p$ as a random variable, $p \in [0, 1]$.

- firstly we will assume $p$ is uniformly distributed, $p \sim \text{Uniform}(0, 1) \Rightarrow P(p)=1$ (since we have no prior beliefs about the coin, every $p$ is equally likely).

- using binomial distribution with $n=10, k=7$, probability for observed data given $p$ is $P(\text{data} | p) = \binom{10}{7}p^7(1-p)^3$

- using Bayes' theorem and conditional probability, we can find probability of $p$ given data: $P(p | \text{data} ) \propto  P(\text{data} | p) \cdot P(p) = p^7(1-p)^3$

This expression is the unnormalized Beta distribution with $\alpha = 8$ and $\beta = 4$, Beta(8,4) $= p^{\alpha - 1} (1-p)^{\beta - 1}$

For the PDF to be normalized we divide by the integral (Beta function) $\rightarrow \text{Beta}(p; \alpha, \beta) = \frac{p^{\alpha - 1} (1-p)^{\beta - 1}}{B(\alpha, \beta)}$

The curve of the PDF $p^7(1-p)^3$ has a peak (mode) at $p=0.7$ which indicates the value $p=0.7$ is the most likely (not most probable!) value for the coin given the data. It aligns with the classical probability approach.

```python
x = np.linspace(0, 1, 100)
pdf = x**7*(1-x)**3
plt.plot(x, pdf)
plt.show()
```

```python
from scipy.stats import beta

x = np.linspace(0, 1, 100)
b = beta.pdf(x, 8, 4)
plt.plot(x, b)

peak_x = 0.7
peak_y = beta.pdf(0.7, 8, 4)

plt.scatter(peak_x, peak_y, color='red')
plt.vlines(peak_x, 0, peak_y, colors='red', linestyles='dotted')
plt.show()
```
